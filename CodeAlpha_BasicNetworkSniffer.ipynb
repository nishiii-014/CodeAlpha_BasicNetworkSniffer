```python
from scapy.all import sniff, IP, TCP, UDP
from datetime import datetime

packet_count = 0

print("=" * 60)
print("        BASIC NETWORK SNIFFER - CODEALPHA")
print("=" * 60)

def process_packet(packet):

    global packet_count
    packet_count += 1

    print("\n")
    print("=" * 60)
    print(f"Packet Number : {packet_count}")
    print("Time Captured :", datetime.now())
    print("=" * 60)

    # Check if packet contains IP layer
    if packet.haslayer(IP):

        source_ip = packet[IP].src
        destination_ip = packet[IP].dst
        protocol = packet[IP].proto

        print(f"Source IP Address      : {source_ip}")
        print(f"Destination IP Address : {destination_ip}")

        # Protocol Identification
        if packet.haslayer(TCP):
            print("Protocol Used          : TCP")

        elif packet.haslayer(UDP):
            print("Protocol Used          : UDP")

        else:
            print(f"Protocol Number        : {protocol}")

        print(f"Packet Summary         : {packet.summary()}")

    else:
        print("No IP Layer Found")

print("\nStarting Packet Capture...\n")

# Capture 15 packets
sniff(prn=process_packet, count=15)

print("\n")
print("=" * 60)
print("Packet Sniffing Completed Successfully")
print("=" * 60)
```
